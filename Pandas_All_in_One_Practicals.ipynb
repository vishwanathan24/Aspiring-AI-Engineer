## 2. Understanding Series

**What is a Series?**
A Series is like a single column in Excel - it's a list of values with labels (called index).

Think of it like:
```
Index    Value
0    ->   Apple
1    ->   Banana
2    ->   Orange
```

````python
# Creating a Series from a list
fruits = pd.Series(['Apple', 'Banana', 'Orange', 'Mango', 'Grapes'])
print("Series of Fruits:")
print(fruits)
print("\nType:", type(fruits))

# here, 01234 are INTRINSIC INDEX VALUES that is default index value 
````

````python
# Creating a Series with custom index (labels)
# Custom index makes it easier to understand data
prices = pd.Series([100, 50, 75, 120, 60], index=['Apple', 'Banana', 'Orange', 'Mango', 'Grapes'])

print("Fruit Prices:")
print(prices)

# here what you listed in index attribute is called EXTRINSIC INDEX VALUES(user specified index value)
````

````python
car = pd.Series(['virtus','creta','santafe','passat','polo','i20'], index=['vw','hyundai','hyundai','vw','vw','hyundai'])
print(car)
print('only vw cars:',car['vw'])
````

````python
# Series from a dictionary
student_scores = pd.Series({'Alvi': 85, 'Bala': 90, 'Charles': 78, 'Derik': 92})
print("Student Scores:")
print(student_scores)
#print("\nAlvi's score:", student_scores['Alvi'])
````

## 3. Understanding DataFrame

**What is a DataFrame?**
A DataFrame is like an Excel spreadsheet - it has rows AND columns. It's a table with headers and data.

Think of it like:
```
Name      Age    Salary    Department
Alice     28     50000     IT
Bob       32     60000     HR
Charlie   25     45000     IT
```

````python
# Creating a DataFrame from a dictionary
# Each key becomes a column
data = { 'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [28, 32, 25, 30, 27],
    'Salary': [50000, 60000, 45000, 70000, 55000],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'] }

df = pd.DataFrame(data) # feeding dict into DF , here ---> data(dict), df(data frame)
print("Employee Data:")
print(df)
print(f"\nShape: {df.shape}")  # Shape tells us (rows, columns)
#print(f"Type: {type(df)}")
````

````python
# Useful methods to explore data
print("First 2 rows:")
print(df.head(2))

print("----------------")
print("Last 2 rows:")
print(df.tail(2))
print("----------------")

#print("Info about DataFrame:")
#print(df.info())
print("----------------")
      
print("Statistical Summary:")
print(df.describe())
````

## 9. Handling Missing Data

**Real-world data is messy! We need to handle missing values (NaN - Not a Number).**

````python
# Create a DataFrame with missing values
messy_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [28, np.nan, 25, 30, 27],  # Bob's age is missing
    'Salary': [50000, 60000, np.nan, 70000, 55000],  # Charlie's salary is missing
    'Department': ['IT', 'HR', 'IT', np.nan, 'HR']  # Diana's department is missing
}

df_messy = pd.DataFrame(messy_data)
print("DataFrame with Missing Values:")
print(df_messy)
print("\n" + "="*40)
print("\nMissing values count per column:")
print(df_messy.isnull().sum())
````

````python
# ( printing only records with null values)
print("Rows with any missing value:")
print(df_messy[df_messy.isnull().any(axis=1)])
print("\n" + "="*40)
print('-------------------------------------')
````

````python
# printing records with our null values (that is w/o missing values)
print("\nRows without any missing value:")
print(df_messy[df_messy.notnull().all(axis=1)])  # (axis=1) means column
````

````python
# Method 1: Remove rows with missing values
print("Method 1: Remove rows with missing values")
df_clean1 = df_messy.dropna()
print(f"Original shape: {df_messy.shape}")
print(f"After dropna(): {df_clean1.shape}")
print("\nCleaned DataFrame:")
print(df_clean1)
# dropna ---> removes missing value RECORDS in total
````

````python
df_messy
````

````python
# Method 2: Fill missing values with a constant
print("Method 2: Fill missing values")
df_clean2 = df_messy.copy() # deep copy of original dataframe df_messy

df_clean2['Age'].fillna(df_clean2['Age'].mean(), inplace=True)  # Fill with average
df_clean2['Salary'].fillna(0, inplace=True)  # Fill with 0
df_clean2['Department'].fillna('--', inplace=True)  # Fill with 'Unknown'
print("\nAfter filling missing values:")
print(df_clean2)
````

````python
# Method 3: Forward fill (use previous value)
print("Method 3: Forward fill")
df_clean3 = df_messy.copy()

df_clean3.fillna(method='ffill', inplace=True)  # Use previous value

print("\nAfter forward fill:")
print(df_clean3)
````

## 15. Merging DataFrames (Combining data)

**Combining multiple DataFrames together - like VLOOKUP in Excel.**

````python
# Create first DataFrame
employees = pd.DataFrame({
    'EmpID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR']
})

# Create second DataFrame
salaries = pd.DataFrame({
    'EmpID': [1, 2, 3, 4, 5],
    'Salary': [50000, 60000, 45000, 70000, 55000]
})

print("Employees DataFrame:")
print(employees)
print("\nSalaries DataFrame:")
print(salaries)
````

````python
# Merge DataFrames
merged_df = pd.merge(employees, salaries, on='EmpID') # this ON is representing the same JOIN concepts of MYSQL

print("After merging on EmpID:")
print(merged_df)
````

## 16. Concatenating DataFrames (Stacking data)

**Putting DataFrames on top of each other or side by side.**

````python
# Create two DataFrames
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Age': [28, 32]
})

df2 = pd.DataFrame({
    'Name': ['Charlie', 'Diana'],
    'Age': [25, 30]
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
print("\n" + "="*40)

# Concatenate (stack on top of each other)
df_concat = pd.concat([df1, df2],)
print("\nAfter concatenating:")
print(df_concat)
````

## 17. Quick Practice - Real Example

**Let's analyze student grades!**

````python
# Create a student grades dataset
students_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'Math': [85, 92, 78, 88, 90, 76],
    'English': [90, 85, 88, 92, 87, 84],
    'Science': [88, 90, 85, 91, 89, 82],
    'Class': ['A', 'A', 'B', 'A', 'A', 'B']
}

students = pd.DataFrame(students_data)
print("Student Grades:")
print(students)
````

````python
# Task 1: Calculate average grade for each student
students['Average'] = students[['Math', 'English', 'Science']].mean(axis=1)
print("With Average Grade:")
print(students[['Name', 'Math', 'English', 'Science', 'Average']])
````

````python
# Task 2: Find students with average >= 85
print("\nTop Performers (Average >= 85):")
top_students = students[students['Average'] >= 85]
print(top_students[['Name', 'Average', 'Class']])
````

````python
# Task 3: Average grades by Class
print("\nAverage Grades by Class:")
class_avg = students.groupby('Class')[['Math', 'English', 'Science']].mean()
print(class_avg)
````

````python
# Task 4: Who is the top student?
print("\nTop Student:")
top_student = students.loc[students['Average'].idxmax()]
print(f"{top_student['Name']} with average {top_student['Average']:.2f}")

print("\nLowest Performing Student:")
low_student = students.loc[students['Average'].idxmin()]
print(f"{low_student['Name']} with average {low_student['Average']:.2f}")
````

## 18. Summary Cheat Sheet

**Quick reference for important operations:**

````python
print("="*60)
print("PANDAS QUICK INTERVIEW - REFERENCE GUIDE")
print("="*60)

cheatsheet = """
1. CREATE DATA:
   - Series: pd.Series([1, 2, 3])
   - DataFrame: pd.DataFrame({'col1': [1, 2], 'col2': [3, 4]})

2. VIEW DATA:
   - df.head()          -> First 5 rows
   - df.tail()          -> Last 5 rows
   - df.info()          -> Data types and missing values
   - df.describe()      -> Statistical summary
   - df.shape           -> (rows, columns)

3. ACCESS DATA:
   - df['column']       -> Entire column
   - df.loc[0]          -> Entire row by label
   - df.iloc[0]         -> Entire row by position
   - df.loc[0, 'col']   -> Specific cell

4. FILTER DATA:
   - df[df['col'] > 5]  -> Rows where col > 5
   - df[df['col'] == 'value']  -> Rows where col equals value

5. SORT DATA:
   - df.sort_values('col')     -> Ascending
   - df.sort_values('col', ascending=False)  -> Descending

6. STATISTICS:
   - df['col'].mean()   -> Average
   - df['col'].median() -> Middle value
   - df['col'].sum()    -> Total
   - df['col'].min()    -> Minimum
   - df['col'].max()    -> Maximum
   - df['col'].std()    -> Standard deviation

7. GROUP & AGGREGATE:
   - df.groupby('col1')['col2'].mean()  -> Average by group
   - df.groupby('col1').size()          -> Count by group

8. ADD/DROP COLUMNS:
   - df['new_col'] = values    -> Add column
   - df.drop('col', axis=1)    -> Remove column
   - df.rename(columns={...})  -> Rename columns

9. HANDLE MISSING VALUES:
   - df.isnull().sum()         -> Count missing values
   - df.dropna()               -> Remove rows with missing values
   - df.fillna(value)          -> Fill missing with value

10. READ/WRITE:
    - pd.read_csv('file.csv')      -> Read CSV
    - df.to_csv('file.csv')        -> Write CSV
    - pd.read_excel('file.xlsx')   -> Read Excel
    - df.to_excel('file.xlsx')     -> Write Excel

11. MERGE/CONCAT:
    - pd.merge(df1, df2, on='col')  -> Join like SQL
    - pd.concat([df1, df2])         -> Stack DataFrames
"""

print(cheatsheet)
print("="*60)
````

## Interactive Exercise

**Try it yourself! Analyze this sales data:**

````python
# Sales data
sales_data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Region': ['North', 'North', 'South', 'South', 'North', 'South', 'North', 'South'],
    'Quantity': [5, 15, 12, 8, 7, 10, 6, 9],
    'Price': [1000, 25, 75, 300, 1000, 25, 75, 300]
}

sales = pd.DataFrame(sales_data)
print("Sales Data:")
print(sales)
````

````python
# Exercise 1: Calculate total revenue for each row
sales['Revenue'] = sales['Quantity'] * sales['Price']
print("\nWith Revenue Column:")
print(sales)
print(f"\nTotal Revenue: ${sales['Revenue'].sum():,.2f}")
````

````python
# Exercise 2: Revenue by Product
print("\nRevenue by Product:")
product_revenue = sales.groupby('Product')['Revenue'].sum().sort_values(ascending=False)
print(product_revenue)
````

````python
# Exercise 3: Revenue by Region
print("\nRevenue by Region:")
region_revenue = sales.groupby('Region')['Revenue'].sum()
print(region_revenue)
print(f"\nHighest Revenue Region: {region_revenue.idxmax()} with ${region_revenue.max():,.2f}")
````

````python
# Exercise 4: Complex aggregation
print("\nDetailed Analysis by Region and Product:")
analysis = sales.groupby(['Region', 'Product']).agg({
    'Quantity': 'sum',
    'Revenue': 'sum',
    'Price': 'first'
}).reset_index()
print(analysis)
print("\nSorted by Revenue (highest first):")
print(analysis.sort_values('Revenue', ascending=False))
````
